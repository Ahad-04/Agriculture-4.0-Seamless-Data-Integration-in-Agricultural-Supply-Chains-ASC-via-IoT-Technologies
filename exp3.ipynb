```python
import pandas as pd
from datetime import timedelta

# File path
file_path = r"D:\ChipVista\Projects\decaying fruits\EXP_03_NANO.csv"

# Read CSV file (auto-detect commas or tabs)
df = pd.read_csv(file_path)

# Create new time values (each 10 minutes apart, reset every 24 hours)
new_times = []
for i in range(len(df)):
    minutes = (10 * i) % 1440  # 1440 minutes = 24 hours
    h = minutes // 60
    m = minutes % 60
    new_times.append(f"{h:02d}:{m:02d}:00")

# Replace only the first column (keep others untouched)
df.iloc[:, 0] = new_times

# Save updated CSV (same columns preserved)
output_path = r"D:\ChipVista\Projects\decaying fruits\EXP_03_NANO_updated.csv"
df.to_csv(output_path, index=False)

print("✅ Only 'Time' column replaced — all other data preserved!")
print(f"Saved file: {output_path}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import os

# ==== File Paths ====
file_path = r"D:\ChipVista\Projects\decaying fruits\EXP_03_NANO_updated.csv"
save_folder = os.path.dirname(file_path)  # same folder for saving images

# ==== Read CSV ====
df = pd.read_csv(file_path)
df.columns = df.columns.str.strip()
print("Columns in CSV:", df.columns.tolist())

# ==== Clean data ====
for col in ["mq2_GI", "mq4_GI"]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()
        df[col] = pd.to_numeric(df[col], errors="coerce")

# ==== Generate datetime values (starting from 18/10/2025, 16:00:00) ====
start_datetime = datetime(2025, 10, 18, 0, 0, 0)
datetimes = [start_datetime + timedelta(minutes=10 * i) for i in range(len(df))]
df["Datetime"] = datetimes


# ==== Plot 1: MQ2 ====
plt.figure(figsize=(10, 5))
plt.plot(df["Datetime"], df["mq2_GI"], marker="o", linewidth=2)
plt.title("MQ2 GI vs Time")
plt.xlabel("Time (starting from 18/10/2025)")
plt.ylabel("MQ2 GI")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

mq2_path = os.path.join(save_folder, "MQ2_EXP3.png")
plt.savefig(mq2_path, dpi=300)
print(f"✅ MQ2 graph saved at: {mq2_path}")
plt.close()

# ==== Plot 2: MQ4 ====
plt.figure(figsize=(10, 5))
plt.plot(df["Datetime"], df["mq4_GI"], color="orange", marker="o", linewidth=2)
plt.title("MQ4 GI vs Time")
plt.xlabel("Time (starting from 18/10/2025)")
plt.ylabel("MQ4 GI")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

mq4_path = os.path.join(save_folder, "MQ4_EXP3.png")
plt.savefig(mq4_path, dpi=300)
print(f"✅ MQ4 graph saved at: {mq4_path}")
plt.close()

print("🎯 Both graphs generated and saved successfully!")
```
