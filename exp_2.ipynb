```python
import pandas as pd
from datetime import timedelta

# File path
file_path = r"D:\ChipVista\Projects\decaying fruits\EXP_02_UNO.csv"

# Read CSV file (auto-detect commas or tabs)
df = pd.read_csv(file_path)

# Create new time values (each 10 minutes apart, reset every 24 hours)
new_times = []
for i in range(len(df)):
    minutes = (10 * i) % 1440  # 1440 minutes = 24 hours
    h = minutes // 60
    m = minutes % 60
    new_times.append(f"{h:02d}:{m:02d}:00")

# Replace only the first column (keep others untouched)
df.iloc[:, 0] = new_times

# Save updated CSV (same columns preserved)
output_path = r"D:\ChipVista\Projects\decaying fruits\EXP_02_UNO_updated.csv"
df.to_csv(output_path, index=False)

print("✅ Only 'Time' column replaced — all other data preserved!")
print(f"Saved file: {output_path}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import os

# ==== File Paths ====
file_path = r"D:\ChipVista\Projects\decaying fruits\EXP_02_UNO_updated.csv"
save_folder = os.path.dirname(file_path)  # same folder for saving images


# ==== Read CSV ====
df = pd.read_csv(file_path)
print("Columns in CSV:", df.columns.tolist())


# ==== Generate datetime values (starting from 17/10/2025, 17:00:00) ====
start_datetime = datetime(2025, 10, 17, 17, 0, 0)
datetimes = [start_datetime + timedelta(minutes=10 * i) for i in range(len(df))]
df["Time"] = datetimes

# ==== Plot 1: MQ2 ====
plt.figure(figsize=(10, 5))
plt.plot(df["Time"], df["mq2_ppm"], marker="o", linewidth=2)
plt.title("MQ2 PPM vs Time")
plt.xlabel("Time (starting from 17/10/2025)")
plt.ylabel("MQ2 PPM")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

mq2_path = os.path.join(save_folder, "MQ2_EXP2.png")
plt.savefig(mq2_path, dpi=300)
print(f"✅ MQ2 graph saved at: {mq2_path}")
plt.close()

# ==== Plot 2: MQ4 ====
plt.figure(figsize=(10, 5))
plt.plot(df["Time"], df["mq4_ppm"], color="orange", marker="o", linewidth=2)
plt.title("MQ4 PPM vs Time")
plt.xlabel("Time (starting from 17/10/2025)")
plt.ylabel("MQ4 PPM")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

mq4_path = os.path.join(save_folder, "MQ4_EXP2.png")
plt.savefig(mq4_path, dpi=300)
print(f"✅ MQ4 graph saved at: {mq4_path}")
plt.close()

print("🎯 Both graphs generated and saved successfully!")
```
